<h1>Payment Gateway</h1>

```python
class CreditCard:
    def pay(self):
        print("Processing Credit Card Payment")


class UPI:
    def pay(self):
        print("Processing UPI Payment")


class NetBanking:
    def pay(self):
        print("Processing Net Banking Payment")


class WalletPayment:
    def pay(self):
        print("Processing Wallet Payment")


payments = [
    CreditCard(),
    UPI(),
    NetBanking(),
    WalletPayment()
]

for payment in payments:
    payment.pay()
```
